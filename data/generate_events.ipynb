### Generating Dimension Table Data

#### Importing Packages

```python
import pandas as pd
import numpy as np
import uuid
import random
from datetime import datetime, timedelta
random.seed(42)
np.random.seed(42)
```

#### Declaring variables

```python
no_users = 10000
countries_list = ['USA', 'India', 'UK', 'Australia', 'Canada', 'Germany', 'Denmark', 'Japan', 'Nepal']
start_date = '2025-01-01 00:00:00'
end_date = '2025-01-31 23:59:59'
platform_list = ['iOS','Android','Web']
persona_list = ['power','casual','one-time']
probabilities_persona = [0.2,0.5,0.3]
```

#### Creating unique identifiers for users

```python
user_ids = [str(uuid.uuid4()) for i in range(no_users)]
```

#### Generating date-time and platform using random choices

```python
start = datetime.strptime(start_date, '%Y-%m-%d %H:%M:%S')
end = datetime.strptime(end_date, '%Y-%m-%d %H:%M:%S')
total_seconds = int((end - start).total_seconds())
random_seconds = [random.randrange(total_seconds) for _ in range(no_users)]
signup_datetime = [start + timedelta(seconds=sec) for sec in random_seconds]
platform = random.choices(platform_list,k=no_users)
```

#### Generating persona and country using random choices

```python
persona = np.random.choice(persona_list, size = no_users, p = probabilities_persona)
persona = persona.tolist()
country = random.choices(countries_list, k = no_users)
```

### Creating a datafram with the columns combined

```python
data = {'user_id':user_ids,'signup_ts':signup_datetime,'platform':platform,'persona':persona,'country':country}
df = pd.DataFrame(data)
```
